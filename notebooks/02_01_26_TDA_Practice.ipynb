```python
import matplotlib.pyplot as plt
import gudhi
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

# 1. Create a set of data points by loading the CSV file
# The user wants to load '/content/circle data.csv' and call it 'points'.

df_points = pd.read_csv('/workspaces/Cellular-TDA/data/data sample 1.csv')
points = df_points.to_numpy()

print(f"Loaded {len(points)} points from /workspaces/Cellular-TDA/data/data sample 1.csv")
plt.figure(figsize=(8, 6))
plt.scatter(points[:, 0], points[:, 1], alpha=0.7, color='blue')
plt.title('Scatter Plot of Loaded Points')
plt.xlabel('X-coordinate')
plt.ylabel('Y-coordinate')
plt.grid(True)
plt.gca().set_aspect('equal', adjustable='box')
plt.show()
```

```python
# 2. Build a Rips complex
rips = gudhi.RipsComplex(points=points, max_edge_length=1)

# 3. Create a simplex tree
st = rips.create_simplex_tree(max_dimension=2)

# 4. Compute persistence and store the result
persistence = st.persistence()

# # 5. Extract and print the persistence intervals
# print("Persistence intervals:")
# for dim, (birth, death) in persistence:
#     print(f"Dimension {dim}: [{birth:.4f}, {death:.4f})")

# 6. Plot the persistence diagram
gudhi.plot_persistence_diagram(persistence)
# 7. plot barcode
gudhi.plot_persistence_barcode(persistence)
plt.title("Persistence Barcode")
plt.grid(True)
plt.show()
```
