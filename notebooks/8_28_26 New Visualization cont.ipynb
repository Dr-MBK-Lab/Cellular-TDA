Stategy 2: Averaging components

```python
import io
import matplotlib.pyplot as plt
my_pca_key = {
     
    "condition": "control", 
    "properties": [1] 
}

manager.PCA.run(my_pca_key)

exclusions = {
    "condition": [""],
    "properties": [],
    "embryo_names": ["control2", "control3", "control4"]
}

fig1 = manager.plot_averaged_vectors(
    PCA_key=my_pca_key, 
    do_not_include=exclusions, 
    step=5
)

# fig.set_size_inches(8,5)
# display(fig)

exclusions = {
    "condition": [""],
    "properties": [],
    "embryo_names": ["sox17-GFP_WT_4_3 MMStack_Pos0", "sox17-GFP_WT_4_4 MMStack_Pos0", "sox17-GFP_WT_6_1 MMStack_Pos0"]
}
fig2 = manager.plot_averaged_vectors(
    PCA_key=my_pca_key, 
    do_not_include=exclusions, 
    step=5
)
# fig.set_size_inches(8,5)
# display(fig)
def fig_to_array(fig):
    buf = io.BytesIO()
    fig.savefig(buf, format='png', bbox_inches='tight', dpi=150)
    buf.seek(0)
    img_array = plt.imread(buf)
    buf.close()
    return img_array

img1 = fig_to_array(fig1)
img2 = fig_to_array(fig2)

# 2. Close original standalone figures to avoid duplicate pop-ups
plt.close(fig1)
plt.close(fig2)

# 3. Create a master canvas with 1 row and 2 columns
fig_master, (ax_left, ax_right) = plt.subplots(1, 2, figsize=(12, 6))

# 4. Render the images side by side
ax_left.imshow(img1)
ax_left.axis('off')  # Hide pixel coordinates layout axis

ax_right.imshow(img2)
ax_right.axis('off')  # Hide pixel coordinates layout axis

# 5. Show the final canvas
plt.tight_layout()
plt.show()
```
