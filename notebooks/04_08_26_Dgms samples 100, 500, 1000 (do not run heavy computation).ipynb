```python
import time
start_notebook = time.time()
# ... (rest of your first cell)
import numpy as np
import matplotlib.pyplot as plt

# Import TDA utilities
from ripser import Rips
from tadasets import torus, sphere
from persim.landscapes import PersLandscapeExact
from persim.landscapes import plot_landscape_simple
from persim.landscapes import PersistenceLandscaper
from persim import plot_diagrams

# Import Scikit-Learn tools
from sklearn.decomposition import PCA
from sklearn import svm
from sklearn import metrics
from sklearn.model_selection import train_test_split
print(f"Total notebook runtime: {time.time() - start_notebook:.2f} seconds")

```

```python
start_notebook = time.time()

# Instantiate datasets in a loop
all_datasets = {}
sample_sizes = [100, 500, 1000]

for size in sample_sizes:
    all_datasets[f'data_torus_{size}'] = torus(n=size, c=2, a=1)
    all_datasets[f'data_sphere_{size}'] = sphere(n=size, r=2)

# Prepare data for iteration using the dynamically created datasets
datasets_info = [
    (all_datasets[f'data_torus_{sample_sizes[0]}'], f"Torus {sample_sizes[0]} Points", sample_sizes[0]),
    (all_datasets[f'data_sphere_{sample_sizes[0]}'], f"Sphere {sample_sizes[0]} Points", sample_sizes[0]),
    (all_datasets[f'data_torus_{sample_sizes[1]}'], f"Torus {sample_sizes[1]} Points", sample_sizes[1]),
    (all_datasets[f'data_sphere_{sample_sizes[1]}'], f"Sphere { sample_sizes[1]} Points", sample_sizes[1]),
    (all_datasets[f'data_torus_{sample_sizes[2]}'], f"Torus {sample_sizes[2]} Points", sample_sizes[2]),
    (all_datasets[f'data_sphere_{sample_sizes[2]}'], f"Sphere {sample_sizes[2]} Points", sample_sizes[2]),
]
print(f"Total notebook runtime: {time.time() - start_notebook:.2f} seconds")
```

```python
start_notebook = time.time()

# Instantiate simplicial complex
rips_L = Rips(maxdim=2, n_perm= 500)
rips_S = Rips(maxdim=2)
print(f"Total notebook runtime: {time.time() - start_notebook:.2f} seconds")
```

```python
start_notebook = time.time()

# Compute persistence diagrams
dgms_data = {}
for data, title, sample_sizes in datasets_info: 
    if sample_sizes > 400:
        dgms_data[title] = rips_L.fit_transform(data)
    else:
        dgms_data[title] = rips_S.fit_transform(data)
    
print(f"Total notebook runtime: {time.time() - start_notebook:.2f} seconds")
```

```python
start_notebook = time.time()

# Plot persistence diagrams
fig, axs = plt.subplots(2, 3, dpi=150, sharey=True, sharex=True)
fig.set_size_inches(8, 6)
fig.tight_layout(pad=.50)

axs_flat = axs.flatten()

for i, key in enumerate(dgms_data): 
     plot_diagrams(dgms_data[key], ax=axs_flat[i], title= key)

fig.tight_layout()

print(f"Total notebook runtime: {time.time() - start_notebook:.2f} seconds")
```

```python
start_notebook = time.time()

landscapes = {};
for i, key in enumerate(dgms_data): 
    landscapes[key] = {}
    landscapes[key]["H0"] = (PersLandscapeExact(dgms_data[key], hom_deg=0))    
    landscapes[key]["H1"] = (PersLandscapeExact(dgms_data[key], hom_deg=1))
    landscapes[key]["H2"] = (PersLandscapeExact(dgms_data[key], hom_deg=2))
    
print(f"Total notebook runtime: {time.time() - start_notebook:.2f} seconds")
```

```python
# Create figure for persistence landscapes
start_notebook = time.time()


fig_landscapes, axs_landscapes = plt.subplots(6, 3)
fig_landscapes.suptitle("Persistence Landscapes", fontsize=16)
fig_landscapes.set_size_inches(10, 8)
fig_landscapes.tight_layout(pad=2.0)
axs_flat = axs_landscapes.flatten()

number_of_lambdas = 5
for i, key in enumerate(landscapes):
    for j, hom in enumerate(landscapes[key]):
        plot_landscape_simple(
            landscapes[key][hom],
            ax=axs_flat[i*3 + j],
            title= key + " " +  rf"${hom[0]}_{hom[1]}$",
            depth_range= range(number_of_lambdas) 
        )

for pl in axs_landscapes.flatten(): 
    pl.get_legend().remove()

# for ax in axs_flat[::3]:
#     ax.set_xlim(0,1.25)
# for ax in axs_flat[1::3]:
#     ax.set_xlim(0, 2.0)
# for ax in axs_flat[2::3]:
#     ax.set_xlim(0.5, 3.5)


print(f"Total notebook runtime: {time.time() - start_notebook:.2f} seconds")
```
