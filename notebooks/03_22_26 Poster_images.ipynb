```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# We're increasing the resolution_points for a smoother surface.
# Be aware that higher values will increase computation time and memory usage.
resolution_points = 300 # Increased from 100 to 300 for a much smoother surface!

print(f"Generating ultra-smooth sphere and torus surfaces with {resolution_points}x{resolution_points} grid...")

# --- Sphere Generation ---
# A sphere can be defined using spherical coordinates.
u = np.linspace(0, 2 * np.pi, resolution_points)   # Azimuthal angle (0 to 2*pi)
v = np.linspace(0, np.pi, resolution_points)       # Polar angle (0 to pi)

# Create a meshgrid for u and v
U, V = np.meshgrid(u, v)

# Define the radius of the sphere
R_sphere = 1.0

# Convert spherical coordinates to Cartesian coordinates
x_sphere = R_sphere * np.sin(V) * np.cos(U)
y_sphere = R_sphere * np.sin(V) * np.sin(U)
z_sphere = R_sphere * np.cos(V)

# --- Torus Generation ---
# A torus can be defined using parametric equations.
theta = np.linspace(0, 2 * np.pi, resolution_points) # Angle around the tube
phi = np.linspace(0, 2 * np.pi, resolution_points)   # Angle around the torus's central axis

# Create a meshgrid for theta and phi
THETA, PHI = np.meshgrid(theta, phi)

# Define the major radius (R) and minor radius (r) of the torus
R_major = 2.0  # Distance from the center of the hole to the center of the tube
r_minor = 0.8  # Radius of the tube itself

# Parametric equations for a torus
x_torus = (R_major + r_minor * np.cos(THETA)) * np.cos(PHI)
y_torus = (R_major + r_minor * np.cos(THETA)) * np.sin(PHI)
z_torus = r_minor * np.sin(THETA)

# --- Plotting ---
fig = plt.figure(figsize=(12, 6)) # Create a figure with two subplots

# Subplot 1: Sphere
ax1 = fig.add_subplot(121, projection='3d') # 1 row, 2 columns, first plot
ax1.plot_surface(x_sphere, y_sphere, z_sphere, color='skyblue', rstride=1, cstride=1, alpha=0.9,  edgecolor='none')
ax1.set_title(f'Sphere)')
ax1.set_xlabel('X')
ax1.set_ylabel('Y')
ax1.set_zlabel('Z')
ax1.set_box_aspect([np.ptp(x_sphere), np.ptp(y_sphere), np.ptp(z_sphere)]) # Equal aspect ratio

# Subplot 2: Torus
ax2 = fig.add_subplot(122, projection='3d') # 1 row, 2 columns, second plot
ax2.plot_surface(x_torus, y_torus, z_torus, color='lightcoral', rstride=1, cstride=1, alpha=0.9,  edgecolor='none')
ax2.set_title(f'Torus')
ax2.set_xlabel('X')
ax2.set_ylabel('Y')
ax2.set_zlabel('Z')
ax2.set_box_aspect([np.ptp(x_torus), np.ptp(y_torus), np.ptp(z_torus)]) # Equal aspect ratio

plt.tight_layout() # Adjust subplot parameters for a tight layout
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

print("Generating 100 points on sphere and torus surfaces, then adding noise...")

# --- Parameters ---
num_points = 100 # Number of points to generate for each surface
R_sphere = 1.0  # Radius of the sphere

R_major = 2.0   # Major radius of the torus
r_minor = 0.8   # Minor radius of the torus

# Noise scale: This is crucial for making the shapes indistinguishable.
# A scale of 1.0 means the standard deviation of the noise is 1 unit.
# Since the sphere has a radius of 1 and the torus's features are around 0.8-2.8 units,
# this level of noise will significantly blur the original shapes.
noise_scale = 1.0 # Standard deviation of the Gaussian noise

# --- Sphere Point Generation ---
# Generate random spherical coordinates for the sphere
# u: azimuthal angle (0 to 2*pi)
# v: polar angle (0 to pi)
u_sphere_rand = np.random.uniform(0, 2 * np.pi, num_points)
v_sphere_rand = np.random.uniform(0, np.pi, num_points)

# Convert spherical coordinates to Cartesian coordinates
x_sphere_orig = R_sphere * np.sin(v_sphere_rand) * np.cos(u_sphere_rand)
y_sphere_orig = R_sphere * np.sin(v_sphere_rand) * np.sin(u_sphere_rand)
z_sphere_orig = R_sphere * np.cos(v_sphere_rand)

# Stack original coordinates for easier noise addition
sphere_original_points = np.stack([x_sphere_orig, y_sphere_orig, z_sphere_orig], axis=-1)

# Add Gaussian noise
noise_sphere = np.random.normal(loc=0.0, scale=noise_scale, size=sphere_original_points.shape)
cloud_sphere = sphere_original_points + noise_sphere
print(f"Generated {num_points} noisy points for the sphere.")

# --- Torus Point Generation ---
# Generate random parametric coordinates for the torus
# theta: angle around the tube (0 to 2*pi)
# phi: angle around the torus's central axis (0 to 2*pi)
theta_torus_rand = np.random.uniform(0, 2 * np.pi, num_points)
phi_torus_rand = np.random.uniform(0, 2 * np.pi, num_points)

# Parametric equations for a torus
x_torus_orig = (R_major + r_minor * np.cos(theta_torus_rand)) * np.cos(phi_torus_rand)
y_torus_orig = (R_major + r_minor * np.cos(theta_torus_rand)) * np.sin(phi_torus_rand)
z_torus_orig = r_minor * np.sin(theta_torus_rand)

# Stack original coordinates
torus_original_points = np.stack([x_torus_orig, y_torus_orig, z_torus_orig], axis=-1)

# Add Gaussian noise
noise_torus = np.random.normal(loc=0.0, scale=noise_scale, size=torus_original_points.shape)
cloud_torus = torus_original_points + noise_torus


# --- Plotting ---
fig = plt.figure(figsize=(14, 7), dpi=300) 
# Create a figure with two subplots

# Subplot 1: Noisy Sphere Points
ax2 = fig.add_subplot(1,2,1, projection='3d') # 1 row, 2 columns, first plot
ax2.scatter(cloud_sphere[:, 0], cloud_sphere[:, 1], cloud_sphere[:, 2], color='blue', alpha=0.7, s=50, label=f'Point Cloud {num_points} ')
# ax1.set_title(f'Noisy Point Cloud (Sphere Origin)')
ax2.set_xlabel('X')
ax2.set_ylabel('Y')
ax2.set_zlabel('Z')
# Set equal aspect ratio to prevent distortion
ax2.set_box_aspect([1, 1, 1])
ax2.legend(fontsize=18)
ax2.grid(True)

# Subplot 2: Noisy Torus Points
ax1 = fig.add_subplot(122, projection='3d') # 1 row, 2 columns, second plot
ax1.scatter(cloud_torus[:, 0], cloud_torus[:, 1], cloud_torus[:, 2], color='red', alpha=0.7, s=50, label=f'Point Cloud {num_points}')
# ax2.set_title(f'Noisy Point Cloud (Torus Origin)')
ax1.set_xlabel('X')
ax1.set_ylabel('Y')
ax1.set_zlabel('Z')
# Set equal aspect ratio
ax1.set_box_aspect([1, 1, 1])
ax1.legend( fontsize=18)
ax1.grid(True)
# savePic("100 point clouds")
# plt.tight_layout() # Adjust subplot parameters for a tight layout

plt.show()

```

```python
import numpy as np
import matplotlib.pyplot as plt
import gudhi
from gudhi.representations import Landscape

# Disable LaTeX to avoid the dvipng warning
plt.rcParams['text.usetex'] = False


def get_tda_data(cloud):
    rips = gudhi.RipsComplex(points=cloud)
    st = rips.create_simplex_tree(max_dimension=2)
    diag = st.persistence()
    h1_intervals = st.persistence_intervals_in_dimension(1)
    # Filter infinite values for landscape/axis calculations
    h1_clean = h1_intervals[h1_intervals[:, 1] < np.inf] if len(h1_intervals) > 0 else np.array([])
    return diag, h1_clean


diag_t, h1_t = get_tda_data(cloud_torus)


diag_s, h1_s = get_tda_data(cloud_sphere)

# --- CALCULATE GLOBAL LIMITS ---
# Combine all finite intervals to find the "window" of activity
all_finite = []
for d in [diag_t, diag_s]:
    for _, (b, d_val) in d:
        if d_val != np.inf:
            all_finite.append([b, d_val])
all_finite = np.array(all_finite)

f_min = np.min(all_finite)
f_max = np.max(all_finite)
margin = (f_max - f_min) * 0.05
limit_window = [f_min - margin, f_max + margin]

# 3. Plotting 2x3 Grid
fig, axes = plt.subplots(2, 3, figsize=(18, 10))

# --- PLOTTING LOGIC ---
for row, (diag, h1, label) in enumerate([(diag_t, h1_t, "Torus"), (diag_s, h1_s, "Sphere")]):
    # Column 0: Barcode
    gudhi.plot_persistence_barcode(diag, axes=axes[row, 0])
    axes[row, 0].set_xlim(limit_window)
    axes[row, 0].set_title(f"{label}: Barcode")

    # Column 1: Diagram
    gudhi.plot_persistence_diagram(diag, axes=axes[row, 1])
    axes[row, 1].set_xlim(limit_window)
    axes[row, 1].set_ylim(limit_window)
    axes[row, 1].set_title(f"{label}: Diagram ")

    # Column 2: Landscape (H1)
    ls = Landscape(num_landscapes=5, resolution=100, sample_range=[f_min, f_max])
    l_vals = ls.fit_transform([h1]).reshape(5, 100) if len(h1) > 0 else np.zeros((5, 100))
    
    x_values = np.linspace(f_min, f_max, 100)
    for i in range(3):
        axes[row, 2].plot(x_values, l_vals[i], label=f"Layer {i+1}" if row == 0 else "")
   
    
    axes[row, 2].set_xlim(limit_window)
    axes[row, 2].set_ylim(-0.05, 0.6) # Standardized Y-height for amplitude
    axes[row, 2].set_title(f"{label}: H1 Landscape")
    if row == 0: axes[row, 2].legend()
    

plt.tight_layout()
savePic("All_together.png")
plt.show()
fig = plt.figure(figsize=(14, 7)) 
ax = fig.add_subplot(111) 
gudhi.plot_persistence_barcode(diag_t, axes=ax)
savePic("Barcode.png")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display
from scipy.spatial import cKDTree
import gudhi
from gudhi.representations import Landscape
import os
# --- 1. Generate Vertical Elongated 8-Shape (15 Points) ---
n_points = 15
t = np.linspace(0, 2 * np.pi, n_points, endpoint=False)
points_2d = []
noise_scale = 0.1

for val in t:
    y = 5 * np.sin(val)
    x = 2 * np.sin(val) * np.cos(val)
    x_n = x + np.random.normal(0, noise_scale)
    y_n = y + np.random.normal(0, noise_scale)
    points_2d.append([y_n, x_n])

points_2d = np.array(points_2d)

# --- 2. Geometric Intersection Helper ---
def intersect(a, b, c, d):
    def ccw(A, B, C):
        return (C[1]-A[1]) * (B[0]-A[0]) > (B[1]-A[1]) * (C[0]-A[0])
    if any(np.array_equal(a, x) or np.array_equal(b, x) for x in [c, d]):
        return False
    return ccw(a,c,d) != ccw(b,c,d) and ccw(a,b,c) != ccw(a,b,d)

# --- 3. Visualization Function ---
def update_plot(radius):
    plt.close('all')
    fig, ax = plt.subplots(figsize=(6, 10))
    
    for p in points_2d:
        circ = plt.Circle((p[0], p[1]), radius, color='blue', alpha=0.07, lw=0)
        ax.add_patch(circ)

    tree = cKDTree(points_2d)
    pairs = list(tree.query_pairs(radius))
    dist_pairs = sorted([(np.linalg.norm(points_2d[i] - points_2d[j]), i, j) for i, j in pairs])
    
    valid_edges = []
    # Set for fast lookup of existing connections
    edge_set = set()

    for _, i, j in dist_pairs:
        p1, p2 = points_2d[i], points_2d[j]
        if not any(intersect(p1, p2, points_2d[e1], points_2d[e2]) for e1, e2 in valid_edges):
            valid_edges.append((i, j))
            edge_set.add(tuple(sorted((i, j))))
            ax.plot([p1[0], p2[0]], [p1[1], p2[1]], color='red', lw=1.5, alpha=0.8, zorder=2)

    # --- Triangle Shading Logic ---
    # Loop through all triplets to see if they form a closed triangle
    for i in range(n_points):
        for j in range(i + 1, n_points):
            for k in range(j + 1, n_points):
                # Check if all 3 edges of the triangle (i,j,k) exist in valid_edges
                if (tuple(sorted((i, j))) in edge_set and 
                    tuple(sorted((j, k))) in edge_set and 
                    tuple(sorted((i, k))) in edge_set):
                    
                    # Extract coordinates for the 3 points
                    tri_coords = points_2d[[i, j, k]]
                    ax.fill(tri_coords[:, 0], tri_coords[:, 1], color='cyan', alpha=0.4, zorder=1)

    ax.scatter(points_2d[:,0], points_2d[:,1], c='darkblue', s=40, zorder=10)
    ax.set_aspect('equal')
    ax.set_xlim(-7, 7)
    ax.set_ylim(-4, 4)
    ax.set_title(f"Figure Eight: Radius {radius:.2f}")
    plt.grid(True, linestyle='--', alpha=0.6)
    savePic("twoloops.png")
    plt.show()

# --- 4. Interface ---
radius_slider = widgets.FloatSlider(
    value=1.5, min=0.1, max=6.0, step=0.1, 
    description='Growth Radius:',
    style={'description_width': 'initial'}
)
output = widgets.interactive_output(update_plot, {'radius': radius_slider})
display(radius_slider, output)

points_2 = points_2d

print(f"Loaded {len(points_2)} points from /content/drive/MyDrive/TDA Project/data sample 2.csv")
plt.figure(figsize=(8, 6))
plt.scatter(points_2[:, 0], points_2[:, 1], alpha=0.7, color='blue')
plt.title('Scatter Plot of Loaded Points')
plt.xlabel('X-coordinate')
plt.ylabel('Y-coordinate')
plt.grid(True)
plt.gca().set_aspect('equal', adjustable='box')
plt.show()

# 2. Build a Rips complex
rips = gudhi.RipsComplex(points=points_2, max_edge_length=6)

# 3. Create a simplex tree
st = rips.create_simplex_tree(max_dimension=2)

# 4. Compute persistence and store the result
persistence_2 = st.persistence()

# # 5. Extract and print the persistence intervals
# print("Persistence intervals:")
# for dim, (birth, death) in persistence:
#     print(f"Dimension {dim}: [{birth:.4f}, {death:.4f})")

# 7. plot barcode
gudhi.plot_persistence_barcode(persistence_2)
plt.title("Persistence Barcode")
savePic("Barcode.png")
# 6. Plot the persistence diagram
gudhi.plot_persistence_diagram(persistence_2)
plt.title("Persistence Diagram")
savePic("Digs.png")
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Circle, Ellipse, FancyArrowPatch
from matplotlib.lines import Line2D

# ------------------------------------------------------------
# Helpers
# ------------------------------------------------------------
def add_glossy_ball(ax, x, y, r, base_color, edgecolor='none', z=3):
    """Draw a glossy sphere-like circle with layered highlights."""
    # main body
    body = Circle((x, y), r, facecolor=base_color, edgecolor=edgecolor, lw=1.0, zorder=z)
    ax.add_patch(body)

    # soft shadow at bottom-right
    for i in range(8):
        rr = r * (0.92 - i * 0.07)
        alpha = 0.04 * (1 - i / 8)
        shadow = Circle((x + r * 0.18, y - r * 0.18), rr,
                        facecolor='black', edgecolor='none', alpha=alpha, zorder=z+0.1)
        ax.add_patch(shadow)

    # bright highlight at upper-left
    for i in range(10):
        rr = r * (0.55 - i * 0.045)
        if rr <= 0:
            break
        alpha = 0.18 * (1 - i / 10)
        hi = Circle((x - r * 0.28, y + r * 0.28), rr,
                    facecolor='white', edgecolor='none', alpha=alpha, zorder=z+0.2)
        ax.add_patch(hi)

    # subtle rim
    rim = Circle((x, y), r, facecolor='none', edgecolor=(0, 0, 0, 0.15), lw=1.5, zorder=z+0.3)
    ax.add_patch(rim)


def add_connection(ax, p1, p2, lw=3.0, color='#2f2f2f', z=2):
    ax.add_line(Line2D([p1[0], p2[0]], [p1[1], p2[1]], lw=lw, color=color, zorder=z))
    # node-like endpoints
    for p in (p1, p2):
        ax.add_patch(Circle(p, 0.006, color='black', zorder=z+0.5))


def add_ring_with_arrows(ax, center, rx, ry, ring_color='#1776d2'):
    """Blue circular loop with white arrows."""
    cx, cy = center

    # outer/inner ellipses to mimic thick ring
    outer = Ellipse((cx, cy), 2*rx, 2*ry, facecolor=ring_color, edgecolor='none', zorder=3)
    inner = Ellipse((cx, cy), 2*rx*0.68, 2*ry*0.68, facecolor='white', edgecolor='none', zorder=4)
    ax.add_patch(outer)
    ax.add_patch(inner)

    # gloss
    gloss = Ellipse((cx - rx*0.08, cy + ry*0.08), 2*rx*0.94, 2*ry*0.94,
                    facecolor='none', edgecolor=(1, 1, 1, 0.16), lw=2.5, zorder=4.5)
    ax.add_patch(gloss)

    # arrows around ring
    angles = np.deg2rad([30, 110, 200, 320])
    for a in angles:
        x1 = cx + np.cos(a) * rx * 0.82
        y1 = cy + np.sin(a) * ry * 0.82
        x2 = cx + np.cos(a + 0.35) * rx * 0.82
        y2 = cy + np.sin(a + 0.35) * ry * 0.82
        arr = FancyArrowPatch((x1, y1), (x2, y2),
                              arrowstyle='-|>', mutation_scale=22,
                              lw=2.5, color='white', zorder=5)
        ax.add_patch(arr)


def add_infinity_tube(ax, center, sx=0.18, sy=0.085, color='#e34a47'):
    """Thick infinity-loop using a parametric curve."""
    cx, cy = center
    t = np.linspace(0, 2*np.pi, 700)
    x = sx * np.sin(t)
    y = sy * np.sin(t) * np.cos(t)

    # main tube
    ax.plot(cx + x, cy + y, color=color, lw=18, solid_capstyle='round', zorder=3)
    # inner highlight
    ax.plot(cx + x - 0.002, cy + y + 0.002, color=(1, 1, 1, 0.18),
            lw=6, solid_capstyle='round', zorder=4)
    # slight dark rim
    ax.plot(cx + x, cy + y, color=(0, 0, 0, 0.12), lw=20, solid_capstyle='round', zorder=2.8)


def add_donut(ax, center, rx=0.11, ry=0.065,
              outer_color='#d8b167', inner_color='#8c6d3d'):
    """Simple torus-like donut."""
    cx, cy = center

    # base body
    outer = Ellipse((cx, cy), 2*rx, 2*ry, facecolor=outer_color, edgecolor='none', zorder=3)
    ax.add_patch(outer)

    # shading layers
    for i in range(7):
        alpha = 0.07 * (1 - i/7)
        e = Ellipse((cx + rx*0.06, cy - ry*0.06), 2*rx*(0.92 - i*0.06), 2*ry*(0.92 - i*0.06),
                    facecolor=(0.55, 0.38, 0.14, alpha), edgecolor='none', zorder=3.2)
        ax.add_patch(e)

    # hole
    hole_outer = Ellipse((cx + rx*0.02, cy + ry*0.02), 2*rx*0.46, 2*ry*0.36,
                         facecolor=inner_color, edgecolor='none', zorder=4)
    hole_inner = Ellipse((cx + rx*0.02, cy + ry*0.02), 2*rx*0.30, 2*ry*0.20,
                         facecolor='white', edgecolor='none', zorder=5)
    ax.add_patch(hole_outer)
    ax.add_patch(hole_inner)

    # glossy highlight
    for i in range(6):
        alpha = 0.08 * (1 - i/6)
        hi = Ellipse((cx - rx*0.22, cy + ry*0.18), 2*rx*(0.34 - i*0.035), 2*ry*(0.24 - i*0.025),
                     facecolor='white', edgecolor='none', alpha=alpha, zorder=5.5)
        ax.add_patch(hi)

    outline = Ellipse((cx, cy), 2*rx, 2*ry, facecolor='none',
                      edgecolor=(0, 0, 0, 0.15), lw=1.5, zorder=6)
    ax.add_patch(outline)


def add_divider(ax, x1, x2, y, color='#8fa1b4', lw=2.0, alpha=0.8):
    ax.add_line(Line2D([x1, x2], [y, y], lw=lw, color=color, alpha=alpha, zorder=1))


# ------------------------------------------------------------
# Canvas
# ------------------------------------------------------------
fig, ax = plt.subplots(figsize=(16, 9), dpi=300)
bg = '#ffffff'
fig.patch.set_facecolor(bg)
ax.set_facecolor(bg)
ax.set_xlim(0, 1)
ax.set_ylim(0, 1)
ax.axis('off')

title_color = '#ffffff'
line_color = '#000000'

# ------------------------------------------------------------
# Title
# ------------------------------------------------------------
ax.text(0.5, 0.92, "Homology Properties",
        ha='center', va='center',
        fontsize=34, color="black",
        fontweight='bold', family='DejaVu Serif')

# add_divider(ax, 0.28, 0.72, 0.875, color=line_color, lw=2)

# ax.text(0.5, 0.83, r"Understanding $H_0$, $H_1$ and $H_2$",
#         ha='center', va='center',
#         fontsize=22, color="black",
#         style='italic', family='DejaVu Serif')

# add_divider(ax, 0.045, 0.955, 0.76, color=line_color, lw=2.2)

# vertical dividers
for x in [1/3, 2/3]:
    ax.add_line(Line2D([x, x], [0.12, 0.73], lw=1.6, color=line_color, alpha=0.8))

# ------------------------------------------------------------
# Section headers
# ------------------------------------------------------------
sections = [
    (1/6, r"$H_0$: Connected Components"),
    (0.5, r"$H_1$: 1-Dimensional Holes"),
    (5/6, r"$H_2$: 2-Dimensional Voids"),
]
for cx, txt in sections:
    ax.text(cx, 0.69, txt,
            ha='center', va='center',
            fontsize=22, color="black",
            fontweight='bold', style='italic', family='DejaVu Serif')
    add_divider(ax, cx - 0.145, cx + 0.145, 0.655, color=line_color, lw=2)

# ------------------------------------------------------------
# LEFT PANEL: H0
# ------------------------------------------------------------
# two red connected nodes
p1 = (0.065, 0.49)
p2 = (0.14, 0.49)
add_connection(ax, p1, p2, lw=2.5)
add_glossy_ball(ax, *p1, 0.015, '#d93034')
add_glossy_ball(ax, *p2, 0.015, '#d93034')

# lone green node
add_glossy_ball(ax, 0.158, 0.37, 0.014, '#61b83f')

# blue graph-like component
nodes = {
    'top':   (0.255, 0.54),
    'left':  (0.205, 0.43),
    'mid':   (0.255, 0.43),
    'right': (0.305, 0.43),
    'bot':   (0.255, 0.32),
}
edges = [
    ('top', 'left'), ('top', 'mid'), ('top', 'right'),
    ('left', 'mid'), ('mid', 'right'),
    ('left', 'bot'), ('mid', 'bot'), ('right', 'bot')
]
for a, b in edges:
    add_connection(ax, nodes[a], nodes[b], lw=2.0)

for k, p in nodes.items():
    add_glossy_ball(ax, p[0], p[1], 0.0145 if k != 'mid' else 0.016, '#2f7fd7')

add_divider(ax, 0.045, 0.31, 0.16, color=line_color, lw=1.6)

ax.text(0.177, 0.085, "Counts connected\ncomponents",
        ha='center', va='center',
        fontsize=20, color="black",
        fontweight='bold', style='italic', family='DejaVu Serif')

# ------------------------------------------------------------
# MIDDLE PANEL: H1
# ------------------------------------------------------------
add_ring_with_arrows(ax, center=(0.44, 0.49), rx=0.065, ry=0.065, ring_color='#1777d4')
add_infinity_tube(ax, center=(0.56, 0.37), sx=0.09, sy=0.065, color='#e14848')

add_divider(ax, 0.36, 0.645, 0.16, color=line_color, lw=1.6)

ax.text(0.5, 0.09, "Counts loops",
        ha='center', va='center',
        fontsize=22, color="black",
        fontweight='bold', style='italic', family='DejaVu Serif')

# ------------------------------------------------------------
# RIGHT PANEL: H2
# ------------------------------------------------------------
add_glossy_ball(ax, 0.74, 0.47, 0.06, '#78b84a')
add_donut(ax, center=(0.885, 0.455), rx=0.07, ry=0.05,
          outer_color='#d8b167', inner_color='#a2814e')

add_divider(ax, 0.69, 0.955, 0.16, color=line_color, lw=1.6)

ax.text(0.822, 0.09, "Counts voids/surfaces",
        ha='center', va='center',
        fontsize=20, color="black",
        fontweight='bold', style='italic', family='DejaVu Serif')

# ------------------------------------------------------------
# Save / show
# ------------------------------------------------------------
plt.tight_layout()
savePic("homology_properties_replica.png")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Polygon, FancyArrowPatch
import os

# ------------------------------------------------------------
# Canvas
# ------------------------------------------------------------
fig, ax = plt.subplots(figsize=(14, 8))
ax.set_xlim(0, 14)
ax.set_ylim(0, 8)
ax.axis("off")
fig.patch.set_facecolor("white")
ax.set_facecolor("white")

# ------------------------------------------------------------
# Style
# ------------------------------------------------------------
LINE = "#4d4d4d"
TEXT = "#5a5a5a"
ARROW = "#7a7a7a"
BLUE = "#79c9d3"
RED = "#d97a7a"
TEAL = "#8bc4c9"
GREEN = "#9bcf7a"

# ------------------------------------------------------------
# Geometry helpers
# ------------------------------------------------------------
def pentagon_points(center, r=0.53):
    """
    Slightly irregular upright pentagon matching the sketch style.
    Vertex order goes around the boundary.
    """
    cx, cy = center
    angles = np.deg2rad([118, 182, 252, 338, 32])
    pts = np.column_stack([cx + r * np.cos(angles), cy + r * np.sin(angles)])
    return pts

def draw_vertices(ax, pts, color=LINE, size=18):
    ax.scatter(pts[:, 0], pts[:, 1], s=size, color=color, zorder=5)

def draw_edges(ax, pts, edges, color=LINE, lw=2.2):
    for i, j in edges:
        ax.plot([pts[i, 0], pts[j, 0]], [pts[i, 1], pts[j, 1]],
                color=color, lw=lw, solid_capstyle="round", zorder=3)

def draw_triangle_fill(ax, pts, tri, color, alpha=0.35, hatch="////"):
    poly = Polygon(
        pts[list(tri)],
        closed=True,
        facecolor=color,
        edgecolor=color,
        linewidth=0.8,
        hatch=hatch,
        alpha=alpha,
        zorder=1
    )
    ax.add_patch(poly)

def boundary_edges(n=5):
    return [(i, (i + 1) % n) for i in range(n)]

def triangle_edges(tri):
    a, b, c = tri
    return [(a, b), (b, c), (a, c)]

def unique_undirected_edges(edges):
    out = []
    seen = set()
    for i, j in edges:
        e = tuple(sorted((i, j)))
        if e not in seen:
            seen.add(e)
            out.append(e)
    return out

def draw_open_chain(ax, center, r=0.53):
    pts = pentagon_points(center, r)
    open_edges = [(0,1), (1,2), (2,3), (3,4)]
    draw_edges(ax, pts, open_edges)
    draw_vertices(ax, pts)
    return pts

def draw_cycle(ax, center, r=0.53):
    pts = pentagon_points(center, r)
    draw_edges(ax, pts, boundary_edges())
    draw_vertices(ax, pts)
    return pts

def draw_cycle_with_fills(ax, center, fills, r=0.53):
    """
    fills: list of (triangle_indices, color)
    Triangles are filled first, then all needed edges are drawn.
    """
    pts = pentagon_points(center, r)

    for tri, color in fills:
        draw_triangle_fill(ax, pts, tri, color)

    edges = boundary_edges()
    for tri, _ in fills:
        edges.extend(triangle_edges(tri))
    edges = unique_undirected_edges(edges)

    draw_edges(ax, pts, edges)
    draw_vertices(ax, pts)
    return pts

def light_blue_arrow(ax, x, y, dx=0.28, dy=0.18):
    arr = FancyArrowPatch(
        (x, y), (x + dx, y + dy),
        arrowstyle="->",
        mutation_scale=14,
        linewidth=1.5,
        color=BLUE,
        zorder=6
    )
    ax.add_patch(arr)

def grey_arrow(ax, x1, y1, x2, y2):
    arr = FancyArrowPatch(
        (x1, y1), (x2, y2),
        arrowstyle="->",
        mutation_scale=16,
        linewidth=1.8,
        color=ARROW,
        zorder=2
    )
    ax.add_patch(arr)

# ------------------------------------------------------------
# Top algebraic sequence
# ------------------------------------------------------------
top_y = 6.85
xs = [0.9, 3.0, 5.0, 7.0, 9.0, 11.0, 13.0]

ax.text(xs[0], top_y, "0", ha="center", va="center",
        fontsize=30, color=TEXT)

for x in xs[1:-1]:
    ax.text(x, top_y, r"$\mathbb{Z}_2$", ha="center", va="center",
            fontsize=32, color=TEXT)

ax.text(xs[-1], top_y, "0", ha="center", va="center",
        fontsize=30, color=TEXT)

for a, b in zip(xs[:-1], xs[1:]):
    grey_arrow(ax, a + 0.38, top_y, b - 0.38, top_y)

# ------------------------------------------------------------
# Middle diagram row
# ------------------------------------------------------------
shape_y = 4.9
shape_xs = [1.0, 3.0, 5.0, 7.0, 9.1, 11.25]

# Vertex numbering around the pentagon:
# 0 = upper-left
# 1 = left
# 2 = lower-left
# 3 = lower-right
# 4 = upper-right

# 1) no cycle
draw_open_chain(ax, (shape_xs[0], shape_y))

# 2) cycle forms
draw_cycle(ax, (shape_xs[1], shape_y))

# 3) cycle persists (one triangle)
draw_cycle_with_fills(
    ax, (shape_xs[2], shape_y),
    fills=[((0, 1, 4), RED)]
)

# 4) cycle persists (same class)
draw_cycle_with_fills(
    ax, (shape_xs[3], shape_y),
    fills=[((0, 1, 4), RED)]
)

# 5) cycle persists (two triangles sharing only one edge, no overlap)
draw_cycle_with_fills(
    ax, (shape_xs[4], shape_y),
    fills=[
        ((0, 1, 4), TEAL),
        ((1, 2, 3), RED),
    ]
)

# 6) cycle dies (three adjacent triangles, edge-sharing only, no overlap)
draw_cycle_with_fills(
    ax, (shape_xs[5], shape_y),
    fills=[
        ((0, 1, 4), TEAL),   # top-left / top-right
        ((1, 2, 3), GREEN),  # lower-left band
        ((1, 3, 4), RED),    # right-side triangle
    ]
)

# little blue arrows above the evolving shapes
for x in shape_xs[1:]:
    light_blue_arrow(ax, x + 0.08, shape_y + 0.48)

# ------------------------------------------------------------
# Labels under shapes
# ------------------------------------------------------------
ax.text(shape_xs[0], 4.12, "no loop", ha="center", va="center",
        fontsize=22, color=TEXT)

ax.text(shape_xs[1], 3.98, "Loop\nforms", ha="center", va="center",
        fontsize=22, color=TEXT, linespacing=0.9)

ax.text(shape_xs[5], 4.04, "loop\ndies", ha="center", va="center",
        fontsize=22, color=TEXT, linespacing=0.9)

# "cycle persists" underline arrow
ax.plot([4.78, 8.55], [3.58, 3.58], color=ARROW, lw=1.6, zorder=2)
grey_arrow(ax, 5.1, 3.53, 8.52, 3.53)
ax.text(6.68, 3.36, "loop persists", ha="center", va="center",
        fontsize=30, color=TEXT)

# ------------------------------------------------------------
# Bottom notes
# ------------------------------------------------------------
ax.text(
    0.48, 2.15,
    "-  Sequence of homology groups is made up\n   of sequences like this.",
    ha="left", va="center",
    fontsize=21, color=TEXT
)

ax.text(
    0.48, 1.02,
    "-  These " + r"$\mathbb{Z}_2$" + " sequences are essentially unique",
    ha="left", va="center",
    fontsize=21, color=TEXT
)

plt.tight_layout()
savePic("replicated_homology_sequence_no_overlap.png")
plt.show()

# To save:
```

```python
import numpy as np
import plotly.graph_objects as go
from scipy.spatial import Delaunay
import ipywidgets as widgets

# --- 1. Generate Torus Points ---
n_points = 200
R, r = 5, 2
theta = np.random.uniform(0, 2 * np.pi, n_points)
phi = np.random.uniform(0, 2 * np.pi, n_points)
x = (R + r * np.cos(phi)) * np.cos(theta)
y = (R + r * np.cos(phi)) * np.sin(theta)
z = r * np.sin(phi)
points_3d = np.stack([x, y, z], axis=1)

def get_sphere_geometry(radius, res=20):
    """Calculates smooth vertices and faces for a single sphere."""
    u, v = np.mgrid[0:2*np.pi:complex(res), 0:np.pi:complex(res)]
    xs, ys, zs = (radius * np.cos(u)*np.sin(v)).flatten(), (radius * np.sin(u)*np.sin(v)).flatten(), (radius * np.cos(v)).flatten()
    
    i_list, j_list, k_list = [], [], []
    for r_idx in range(res - 1):
        for c_idx in range(res - 1):
            p1, p2 = r_idx * res + c_idx, r_idx * res + (c_idx + 1)
            p3, p4 = (r_idx + 1) * res + c_idx, (r_idx + 1) * res + (c_idx + 1)
            i_list.extend([p1, p1]); j_list.extend([p2, p3]); k_list.extend([p3, p4])
    return xs, ys, zs, i_list, j_list, k_list

def update_plot(radius):
    tri = Delaunay(points_3d)
    valid_tri_i, valid_tri_j, valid_tri_k = [], [], []
    
    for s in tri.simplices:
        for f in [(s[0], s[1], s[2]), (s[0], s[1], s[3]), (s[0], s[2], s[3]), (s[1], s[2], s[3])]:
            pts = points_3d[list(f)]
            if all(np.linalg.norm(pts[a]-pts[b]) <= radius for a, b in [(0,1), (1,2), (2,0)]):
                valid_tri_i.append(f[0]); valid_tri_j.append(f[1]); valid_tri_k.append(f[2])

    fig = go.Figure()

    # 1. Tiled Faint Blue Spheres (Alpha=0.01 Style)
    sx, sy, sz, si, sj, sk = get_sphere_geometry(radius, res=20)
    all_x, all_y, all_z, all_i, all_j, all_k = [], [], [], [], [], []
    
    for p in points_3d:
        offset = len(all_x)
        all_x.extend(sx + p[0]); all_y.extend(sy + p[1]); all_z.extend(sz + p[2])
        all_i.extend([idx + offset for idx in si])
        all_j.extend([idx + offset for idx in sj])
        all_k.extend([idx + offset for idx in sk])

    fig.add_trace(go.Mesh3d(
        x=all_x, y=all_y, z=all_z, i=all_i, j=all_j, k=all_k, 
        opacity=0.01, # Extremely faint per your snippet
        color='skyblue', hoverinfo='skip', showlegend=False
    ))

    # 2. Solid Black Torus Surface
    if valid_tri_i:
        fig.add_trace(go.Mesh3d(
            x=points_3d[:,0], y=points_3d[:,1], z=points_3d[:,2], 
            i=valid_tri_i, j=valid_tri_j, k=valid_tri_k, 
            color='blue', 
            opacity=1.0, 
            name='Connection Mesh'
        ))

    # 3. Dark Blue Markers (s=20, alpha=0.8 Style)
    fig.add_trace(go.Scatter3d(
        x=points_3d[:,0], y=points_3d[:,1], z=points_3d[:,2], 
        mode='markers', 
        marker=dict(size=4, color='darkblue', opacity=0.8),
        name='Noisy Points'
    ))

    fig.update_layout(
        scene=dict(
            aspectmode='data',
            xaxis=dict(visible=True, showgrid=True),
            yaxis=dict(visible=True, showgrid=True),
            zaxis=dict(visible=True, showgrid=True)
        ), 
        height=700, margin=dict(t=30), title=f"Matplotlib Styled Spheres | Connection Radius: {radius:.2f}"
    )
    fig.show()

widgets.interact(update_plot, radius=widgets.FloatSlider(value=2.0, min=0.5, max=5.0, step=0.1, continuous_update=False))
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d.art3d import Poly3DCollection
from scipy.spatial import Delaunay
from scipy.spatial.distance import pdist, squareform
import cv2
import os

# --- Configuration ---
n_points = 100
total_frames = 200
max_eps = 4.0
R, r = 5, 2
output_folder = "tda_frames"
if not os.path.exists(output_folder): os.makedirs(output_folder)

# --- 1. Generate Points & Distance Matrix ---
theta = np.random.uniform(0, 2 * np.pi, n_points)
phi = np.random.uniform(0, 2 * np.pi, n_points)
x_pts = (R + r * np.cos(phi)) * np.cos(theta)
y_pts = (R + r * np.cos(phi)) * np.sin(theta)
z_pts = r * np.sin(phi)
points_3d = np.stack([x_pts, y_pts, z_pts], axis=1)

# Pre-calculate distances for the barcode
dist_matrix = squareform(pdist(points_3d))
all_edge_births = np.sort(dist_matrix[np.triu_indices(n_points, k=1)])
# We'll just show the first 50 persistent features for clarity
n_bars = 50 
bar_births = all_edge_births[:n_bars]

# --- 2. Rendering ---
fig = plt.figure(figsize=(16, 8))
ax_3d = fig.add_subplot(121, projection='3d')
ax_bar = fig.add_subplot(122)

for frame in range(total_frames):
    ax_3d.clear()
    ax_bar.clear()
    
    current_eps = (frame / total_frames) * max_eps
    rot_angle = (frame / total_frames) * 360

    # A. 3D Simplicial Complex (Left)
    # Drawing edges that have "born" by current_eps
    adj = dist_matrix <= current_eps
    for i in range(n_points):
        for j in range(i + 1, n_points):
            if adj[i, j]:
                ax_3d.plot([x_pts[i], x_pts[j]], [y_pts[i], y_pts[j]], [z_pts[i], z_pts[j]], 
                           color='red', alpha=0.3, lw=0.5)
    
    ax_3d.scatter(x_pts, y_pts, z_pts, c='darkblue', s=15)
    ax_3d.set_xlim(-8, 8); ax_3d.set_ylim(-8, 8); ax_3d.set_zlim(-4, 4)
    ax_3d.view_init(elev=30, azim=rot_angle)
    ax_3d.set_title(f"Vietoris-Rips Complex (ε = {current_eps:.2f})")

    # B. Persistence Barcode (Right)
    # Bars grow from their 'birth' time to the 'current' time
    for i, birth in enumerate(bar_births):
        if current_eps > birth:
            ax_bar.hlines(i, birth, current_eps, colors='teal', lw=2)

    ax_bar.set_xlim(0, max_eps)
    ax_bar.set_ylim(-1, n_bars)
    ax_bar.set_title("Persistence Barcode ($H_1$ Features)")
    ax_bar.set_xlabel("Filtration Parameter (ε)")
    ax_bar.set_yticks([]) # Hide Y-axis as ordering is arbitrary

    plt.savefig(f"{output_folder}/frame_{frame:03d}.png")
    if frame % 20 == 0: print(f"Progress: {frame}/{total_frames}")

plt.close()
img_list = sorted([f for f in os.listdir(output_folder) if f.endswith('.png')])
frame = cv2.imread(os.path.join(output_folder, img_list[0]))
height, width, layers = frame.shape
video = cv2.VideoWriter('v-ripsplusDiagram.mp4', cv2.VideoWriter_fourcc(*'mp4v'), 10, (width, height))

for img_name in img_list:
    video.write(cv2.imread(os.path.join(output_folder, img_name)))

video.release()
print("Video saved as rips_filtration.mp4")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d.art3d import Poly3DCollection
from scipy.spatial import Delaunay
from scipy.spatial.distance import pdist, squareform

# --- Configuration & Point Generation ---
n_points = 80
R, r = 5, 2
theta = np.random.uniform(0, 2 * np.pi, n_points)
phi = np.random.uniform(0, 2 * np.pi, n_points)
points_3d = np.stack([
    (R + r * np.cos(phi)) * np.cos(theta),
    (R + r * np.cos(phi)) * np.sin(theta),
    r * np.sin(phi)
], axis=1)

# Pre-calculate distances and triangulation
dist_matrix = squareform(pdist(points_3d))
tri = Delaunay(points_3d)

# Simulated H1 Persistence (2 main torus loops + noise)
h1_features = [(1.2, 3.2), (1.5, 3.0)] + [(np.random.uniform(0.5, 2.0), np.random.uniform(2.1, 2.5)) for _ in range(8)]

def render_frame(current_eps, angle):
    fig = plt.figure(figsize=(16, 8))
    ax_3d = fig.add_subplot(121, projection='3d')
    ax_bar = fig.add_subplot(122)

    # --- 1. Simplicial Complex with Skyblue Triangles ---
    faces = []
    for s in tri.simplices:
        # A triangle is added when all its edges are <= current_eps
        p_f = points_3d[s[:3]]
        dists = [np.linalg.norm(p_f[i]-p_f[j]) for i,j in [(0,1),(1,2),(2,0)]]
        if max(dists) <= current_eps:
            faces.append(p_f)
            
    if faces:
        poly = Poly3DCollection(faces, facecolors='skyblue', edgecolors='red', alpha=0.5, linewidths=0.2)
        ax_3d.add_collection3d(poly)
    
    ax_3d.scatter(points_3d[:,0], points_3d[:,1], points_3d[:,2], c='darkblue', s=10)
    ax_3d.set_title(f"Torus Filtration (ε = {current_eps:.2f})")
    ax_3d.view_init(elev=30, azim=angle)
    ax_3d.set_axis_off()

    # --- 2. H1 Persistence Barcode ---
    for i, (birth, death) in enumerate(h1_features):
        if current_eps > birth:
            # Bar grows until it reaches its death or the current epsilon
            active_end = min(current_eps, death)
            ax_bar.hlines(i, birth, active_end, colors='teal', lw=3)
            if current_eps >= death:
                ax_bar.scatter(death, i, color='red', s=15, zorder=5) # Death point

    ax_bar.set_xlim(0, 4.0); ax_bar.set_ylim(-1, len(h1_features))
    ax_bar.set_title("$H_1$ Persistence Barcode (Torus Loops)")
    ax_bar.set_xlabel("Filtration Parameter (ε)")
    plt.show()

# Example: Render a single frame midway through the filtration
render_frame(current_eps=2.2, angle=95)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d.art3d import Poly3DCollection
from scipy.spatial import Delaunay
from scipy.spatial.distance import pdist, squareform
import imageio.v2 as imageio
import os

# --- Configuration ---
n_points = 100
total_frames = 200
max_eps = 5.0
R, r = 5, 2
output_folder = "tda_pure_torus"
if not os.path.exists(output_folder):
    os.makedirs(output_folder)

# --- 1. Generate Points ---
theta = np.random.uniform(0, 2 * np.pi, n_points)
phi = np.random.uniform(0, 2 * np.pi, n_points)
points_3d = np.stack([
    (R + r * np.cos(phi)) * np.cos(theta),
    (R + r * np.cos(phi)) * np.sin(theta),
    r * np.sin(phi)
], axis=1)

tri = Delaunay(points_3d)

# H1 barcode: TWO primary loops + noise
h1_bars = [(0.8, 3.3), (1.2, 3.0)] + \
          [(np.random.uniform(0.4, 1.1), np.random.uniform(1.2, 1.5)) for _ in range(35)]

# UPDATED LABELS
labels = {0: "Major loop", 1: "Minor loop"}

# --- 2. Rendering ---
frame_paths = []
for frame_idx in range(total_frames):
    fig = plt.figure(figsize=(18, 6), dpi=150)
    ax_3d = fig.add_subplot(131, projection='3d')
    ax_bar = fig.add_subplot(132)
    ax_diag = fig.add_subplot(133)

    current_eps = (frame_idx / total_frames) * max_eps
    rot_angle = (frame_idx / total_frames) * 360

    # A. 3D Torus Plot
    faces = []
    for s in tri.simplices:
        for f in [(0,1,2), (0,1,3), (0,2,3), (1,2,3)]:
            face_pts = points_3d[s[list(f)]]
            dists = [np.linalg.norm(face_pts[i]-face_pts[j]) for i,j in [(0,1),(1,2),(2,0)]]
            if max(dists) <= current_eps:
                faces.append(face_pts)

    if faces:
        poly = Poly3DCollection(faces, facecolors='skyblue', edgecolors='red', alpha=0.2, linewidths=0.1)
        ax_3d.add_collection3d(poly)

    ax_3d.scatter(points_3d[:,0], points_3d[:,1], points_3d[:,2], c='darkblue', s=5, alpha=0.7)
    ax_3d.set_xlim(-8, 8); ax_3d.set_ylim(-8, 8); ax_3d.set_zlim(-4, 4)
    ax_3d.view_init(elev=45, azim=rot_angle)
    ax_3d.set_axis_off() 
    ax_3d.set_title(f"Torus ε = {current_eps:.2f}")

    # B. Persistence Barcode
    for i, (birth, death) in enumerate(h1_bars):
        if current_eps > birth:
            end_val = min(current_eps, death)
            color = 'crimson' if i < 2 else 'teal'
            # Added zorder to lines
            ax_bar.hlines(i, birth, end_val, colors=color, lw=2.5 if i < 2 else 0.5, alpha=0.8, zorder=1)
            
            # UPDATED TEXT: BLACK COLOR AND ON TOP (zorder)
            if i in labels:
                ax_bar.text(birth, i + 0.5, labels[i], color='black', fontsize=9, fontweight='bold', zorder=10)
    
    ax_bar.set_xlim(0, 4) 
    ax_bar.set_ylim(-1, len(h1_bars))
    ax_bar.set_title("Barcode ($H_1$ Features)")
    ax_bar.set_yticks([])

    # C. Persistence Diagram
    ax_diag.plot([0, max_eps], [0, max_eps], 'k--', alpha=0.3)
    for i, (birth, death) in enumerate(h1_bars):
        if current_eps >= birth:
            color = 'crimson' if i < 2 else 'teal'
            ax_diag.scatter(birth, min(current_eps, death), color=color, s=25, alpha=0.7)
    
    ax_diag.set_xlim(0, 4)
    ax_diag.set_ylim(0, max_eps)
    ax_diag.set_title("Birth vs Death")
    ax_diag.grid(True, linestyle=':', alpha=0.5)

    plt.tight_layout()
    path = f"{output_folder}/f_{frame_idx:03d}.png"
    plt.savefig(path)
    frame_paths.append(path)
    plt.close()

# --- 3. Save as GIF & Cleanup ---
images = [imageio.imread(f) for f in frame_paths]
imageio.mimsave('torus_pure_tda.gif', images, fps=12, loop=0)

for f in frame_paths:
    os.remove(f)
os.rmdir(output_folder)

print("Done! Saved as torus_pure_tda.gif")
```
